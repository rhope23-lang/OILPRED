```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
```

```python
oil_data = pd.read_csv(r'C:\Users\user hope\Documents\OILPRED\DataSet\crude-oil-price.csv')
df = pd.DataFrame(oil_data)
df.head()
```

```python
df.info()
df.describe
```

```python
df.shape
df.columns
```

```python

new_date = df[
    (df['date'] > '1985-01-01') &
    (df['date'] < '2005-01-01')
]
new_date['date'] = pd.to_datetime(new_date['date'])
```

```python
plt.figure(figsize=(10, 5))
plt.plot(new_date['date'], new_date['price'], color='blue', label='Crude Oil Price')
plt.xlabel('Date')
plt.ylabel('Price')
plt.title('Crude Oil Price From 1985 to 2005')
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(10, 5))
df['date'] = pd.to_datetime(df['date'])
plt.plot(df['date'], df['price'], color='blue', label='Crude Oil Price')
plt.xlabel('Date')
plt.ylabel('Price')
plt.title('Crude Oil Price From 1983')
plt.legend()
plt.show()
```
